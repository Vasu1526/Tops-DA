```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

df = pd.read_csv('historical_automobile_sales.csv')
df.head()
```

```python
sales_per_year = df.groupby('Year')['Automobile_Sales'].sum().reset_index()
sales_per_year.plot(x='Year', y='Automobile_Sales', title='Automobile Sales Year Over Year')
plt.ylabel('Total Automobile Sales')
plt.xlabel('Year')
```

```python
sales_by_year_type_recession = df.groupby(['Year', 'Vehicle_Type', 'Recession'])['Automobile_Sales'].sum().reset_index()
for vehicle_type in sales_by_year_type_recession['Vehicle_Type'].unique():
    sub_non_rec = sales_by_year_type_recession[(sales_by_year_type_recession['Vehicle_Type'] == vehicle_type) & (sales_by_year_type_recession['Recession'] == 0)]
    plt.plot(sub_non_rec['Year'], sub_non_rec['Automobile_Sales'], label=f'{vehicle_type} Non-Recession', linestyle='-')
    sub_rec = sales_by_year_type_recession[(sales_by_year_type_recession['Vehicle_Type'] == vehicle_type) & (sales_by_year_type_recession['Recession'] == 1)]
    plt.plot(sub_rec['Year'], sub_rec['Automobile_Sales'], label=f'{vehicle_type} Recession', linestyle='--')
plt.title('Automobile Sales Trends by Vehicle Type and Recession Status')
plt.xlabel('Year')
plt.ylabel('Automobile Sales')
plt.legend()
```

```python
sns.set(style='whitegrid')
bar_data = df.groupby(['Vehicle_Type', 'Recession'])['Automobile_Sales'].mean().reset_index()
sns.barplot(data=bar_data, x='Vehicle_Type', y='Automobile_Sales', hue='Recession')
plt.title('Average Automobile Sales by Vehicle Type: Recession vs Non-Recession')
plt.ylabel('Average Automobile Sales')
plt.xticks(rotation=45)
```

```python
sns.barplot(data=bar_data, x='Vehicle_Type', y='Automobile_Sales', hue='Recession')
plt.title('Comparison of Automobile Sales by Vehicle Type and Recession Status')
plt.ylabel('Average Automobile Sales')
plt.xticks(rotation=45)
```
