```python
# Write a Python function to get the largest number, smallest num and sum of all from the list. 
def list_stats(numbers):
    if not numbers:
        return None, None, 0
    large = max(numbers)
    small = min(numbers)
    total = sum(numbers)
    return large, small, total
numbers = [2, 33, 222, 14, 25]
large, small, total = list_stats(numbers)
print(f"Largest: {large}, Smallest: {small}, Sum: {total}")
```

```python
# Write a Python program to count the number of strings where the string length is 2 or more and the first and last character are same from a given list of strings. 
def match_strings(strings):
    count = 0
    for s in strings:
        if len(s) >= 2 and s[0] == s[-1]:
            count += 1
    return count
strings = ["abc", "aba", "xyz", "aa", "x"]
result = match_strings(strings)
print(f"Number of matching strings: {result}")
```

```python
# Write a Python program to find the second smallest number in a list. 
def second_smallest(numbers):
    unique_numbers = list(set(numbers))
    if len(unique_numbers) < 2:
        return None
    unique_numbers.sort()
    return unique_numbers[1]
numbers = [2, 33, 222, 14, 25]
result = second_smallest(numbers)
if result is not None:
    print(f"Second smallest number: {result}")
else:
    print("Not enough unique numbers to find the second smallest.")
```

```python
# Write a Python function to calculate the factorial of a number (a nonnegative integer) 
def factorial(n):
    if n < 0:
        return "Invalid Number"
    fact = 1
    for i in range(1, n + 1):
        fact *= i
    return fact
n = int(input("Enter a number: "))
result = factorial(n)
print(f"Factorial of {n} is {result}")
```

```python
#  Write a Python function to check whether a number is perfect or not. 
# positive integer which equals the sum of all its divisors
def is_perfect(n):
    if n < 1:
        return False
    divisors_sum = sum(i for i in range(1, n) if n % i == 0)
    return divisors_sum == n
n = int(input("Enter a number: "))
result = is_perfect(n)
if result:
    print(f"{n} is a perfect number.")
else:
    print(f"{n} is not a perfect number.")
```

```python
# Write a Python program to write a list to a file.
def write_lst(lst, file_path):
    try:
        with open(file_path, 'w') as file:
            for item in lst:
                file.write(f"{item}\n")
        print(f"List written {file_path}.")
    except Exception as e:
        print(f"An error occurred: {e}")
lst = [1, 2, 3, 4, 5]
file_path = input("Enter the file path to write the list: ")
write_lst(lst, file_path)
```

```python
# Write a Python program to copy the contents of a file to another file. 
def copy_file_contents(source_path, dest_path):
    try:
        with open(source_path, 'r') as source_file:
            content = source_file.read()
        
        with open(dest_path, 'w') as dest_file:
            dest_file.write(content)
        
        print(f"Contents copied from {source_path} to {dest_path}.")
    except FileNotFoundError:
        print(f"File not found: {source_path}")
    except Exception as e:
        print(f"An error occurred: {e}")
source_path = input("Enter the source file path: ")
dest_path = input("Enter the destination file path: ")
copy_file_contents(source_path, dest_path)
```

```python
# Bar chart showing the total deaths per year for 1984-2016 
import matplotlib.pyplot as plt
import pandas as pd

# Read local CSV file
df = pd.read_csv("U.S.Death.csv")

# Filter years 1984-2016
df = df[df['year'].between(1984, 2016)]

# Count deaths per year
deaths = df['year'].value_counts().sort_index()

# Create bar chart
plt.bar(deaths.index, deaths.values)
plt.xlabel('Year')
plt.ylabel('Deaths')
plt.title('Law Enforcement Deaths (1984-2016)')
```

```python
#  Line chart comparing the yearly deaths in different states 
plt.figure(figsize=(16, 6))
states = df['state'].unique()
for state in states:
    state_data = df[df['state'] == state]
    yearly_deaths = state_data['year'].value_counts().sort_index()
    plt.plot(yearly_deaths.index, yearly_deaths.values, label=state)
plt.xlabel('Year')
plt.ylabel('Deaths')
plt.title('Yearly Deaths by State')
plt.legend()
```
